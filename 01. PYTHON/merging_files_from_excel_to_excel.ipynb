###Gabungin antar hari merged_20250502

```python
# GANTIII AS

import pandas as pd
from google.colab import drive

# Mount Google Drive
drive.mount('/content/drive')

#  GANTIII AS: Path ke file Excel
file1 = '/content/drive/MyDrive/Data PT Check/202505/20250502/01_PO Raja Plafon PVC.xlsx'
file2 = '/content/drive/MyDrive/Data PT Check/202505/20250502/02_Batam po Nawi Jaya 01.05.25.xlsx'
file3 = '/content/drive/MyDrive/Data PT Check/202505/20250502/PO indo bangunan 01.05.25.xlsx'
file4 = '/content/drive/MyDrive/Data PT Check/202505/20250502/PO JANET WALLLBOARD 1 MEI 2025.xlsx'
file5 = '/content/drive/MyDrive/Data PT Check/202505/20250502/PO PT SINAR BINTANG PRATAMA SUKSES 02-05-2025.xlsx'

#  GANTIII AS YG DF1 DAN FILE1 DST. Membaca file dengan dua baris header: baris ke-11 dan 12 (index 10 dan 11 karena 0-based)
df1 = pd.read_excel(file1, header=[10, 11])
df2 = pd.read_excel(file2, header=[10, 11])
df3 = pd.read_excel(file3, header=[10, 11])
df4 = pd.read_excel(file4, header=[10, 11])
df5 = pd.read_excel(file5, header=[10, 11])

# GANTIII AS
#Gabungkan (menumpuk) kedua data
merged_20250502_df = pd.concat([df1, df2, df3, df4, df5], ignore_index=True)

#hapus missing value dari kolom Qty/dus

merged_20250502_df = merged_20250502_df.dropna(subset=[('Qty/dus', 'Unnamed: 7_level_1')])

# Simpan hasil gabungan ke file baru
output_path = '/content/drive/MyDrive/Data PT Check/202505/20250502/merged_20250502.xlsx'
merged_20250502_df.to_excel(output_path)

print("File berhasil digabung dan disimpan sebagai 'merged_20250502.xlsx'")

```

```python
# GANTIII AS. yang diganti nama file 1-n, nama toko, df1-n

import pandas as pd
from google.colab import drive

# Mount Google Drive
drive.mount('/content/drive')

#  GANTIII AS: Path ke file Excel
file1 = '/content/drive/MyDrive/Data PT Check/202505/20250509/art Deco gypsum 9 Mei.xlsx'
file2 = '/content/drive/MyDrive/Data PT Check/202505/20250509/PO Adinjaya Plafon PVC 9 Mei.xlsx'
file3 = '/content/drive/MyDrive/Data PT Check/202505/20250509/PO pijar Jaya 9 mei 2025 wallboard.xlsx'




#  GANTIII AS YG DF1 DAN FILE1 DST. Membaca file dengan dua baris header: baris ke-11 dan 12 (index 10 dan 11 karena 0-based)
df1 = pd.read_excel(file1, header=[10, 11])
df2 = pd.read_excel(file2, header=[10, 11])
df3 = pd.read_excel(file3, header=[10, 11])



#GANTI AS: DF1-DFN

#AS_DF1

##buat kolom "Toko" dari nama file dari df1. --->Yang diganti df1 sampai dfn dan nama filenya, dan tanggal

import pandas as pd
import os

# Path ke file Excel
# GANTIII AS: Ganti path berikut dengan lokasi file Shopee.xlsx yang sebenarnya di Google Drive Anda
file1 = '/content/drive/MyDrive/Data PT Check/202505/20250509/art Deco gypsum 9 Mei.xlsx'

# Baca file Excel
df1 = pd.read_excel(file1, header=[10, 11])

# Ambil nama file (tanpa ekstensi) → misalnya jadi: "Shopee"
nama_toko = os.path.splitext(os.path.basename(file1))[0]

# Tambahkan kolom "Toko"
df1['Toko'] = nama_toko

##AS_BUAT KOLOM TANGGAL
import datetime

# Tambahkan kolom "Tanggal" dengan format datetime
df1['Tanggal'] = datetime.date(2025, 5, 9)

df1['Tanggal'] = pd.to_datetime('2025-05-09')  # format ISO atau bisa juga '05/05/2025'


# AS_DF2

##buat kolom "Toko" dari nama file dari df2

import pandas as pd
import os

# Path ke file Excel
# GANTIII AS: Ganti path berikut dengan lokasi file Shopee.xlsx yang sebenarnya di Google Drive Anda
file2 = '/content/drive/MyDrive/Data PT Check/202505/20250509/PO Adinjaya Plafon PVC 9 Mei.xlsx'

# Baca file Excel
df2 = pd.read_excel(file2, header=[10, 11])

# Ambil nama file (tanpa ekstensi) → misalnya jadi: "Shopee"
nama_toko = os.path.splitext(os.path.basename(file2))[0]

# Tambahkan kolom "Toko"
df2['Toko'] = nama_toko

##AS_BUAT KOLOM TANGGAL
import datetime

# Tambahkan kolom "Tanggal" dengan format datetime
df2['Tanggal'] = datetime.date(2025, 5, 9)

df2['Tanggal'] = pd.to_datetime('2025-05-09')  # format ISO atau bisa juga '05/05/2025'

# AS_df3

##buat kolom "Toko" dari nama file dari df3

import pandas as pd
import os

# Path ke file Excel
# GANTIII AS: Ganti path berikut dengan lokasi file Shopee.xlsx yang sebenarnya di Google Drive Anda
file3 = '/content/drive/MyDrive/Data PT Check/202505/20250509/PO pijar Jaya 9 mei 2025 wallboard.xlsx'

# Baca file Excel
df3 = pd.read_excel(file3, header=[10, 11])

# Ambil nama file (tanpa ekstensi) → misalnya jadi: "Shopee"
nama_toko = os.path.splitext(os.path.basename(file3))[0]

# Tambahkan kolom "Toko"
df3['Toko'] = nama_toko

##AS_BUAT KOLOM TANGGAL
import datetime

# Tambahkan kolom "Tanggal" dengan format datetime
df3['Tanggal'] = datetime.date(2025, 5, 9)

df3['Tanggal'] = pd.to_datetime('2025-05-09')  # format ISO atau bisa juga '05/05/2025'


#AS_FIX MERGE

# GANTIII AS, AS_MERGING DATA
#Gabungkan (menumpuk) kedua data
merged_20250509_df = pd.concat([df1, df2, df3], ignore_index=True)

#hapus missing value dari kolom Qty/dus

merged_20250509_df = merged_20250509_df.dropna(subset=[('Qty/dus', 'Unnamed: 7_level_1')])

# Simpan hasil gabungan ke file baru
output_path = '/content/drive/MyDrive/Data PT Check/202505/20250509/merged_20250509.xlsx'
merged_20250509_df.to_excel(output_path)

print("File berhasil digabung dan disimpan sebagai 'merged_20250509.xlsx'")
```

```python
import pandas as pd
from google.colab import drive
drive.mount('/content/drive')

# Path ke file
file_path = '/content/drive/MyDrive//Data PT Check/202510/20251013/merged_20251013.xlsx'

# Membaca file dengan delimiter tab (\t)
df_merged_20251013 = pd.read_excel(file_path)



##==== MELIHAT GAMBARAN TOKONYA===============
# Menghitung jumlah (count) dan persentase (percentage)
Toko_count = df_merged_20251013['Toko'].value_counts()
Toko_percentage = df_merged_20251013['Toko'].value_counts(normalize=True) * 100


# Menggabungkan dalam satu dataframe
Toko_summary = pd.DataFrame({'Count': Toko_count, 'Percentage': Toko_percentage})


# Menampilkan hasil
print("Ringkasan variabel 'Toko':\n", Toko_summary)
```

```python
##coba lagi 19 juni 18:22

#mainnn
import pandas as pd
import re


# Daftar path file
file_paths = [
    '/content/drive/MyDrive/Data PT Check/202505/merged_202505.xlsx',
    '/content/drive/MyDrive/Data PT Check/202506/merged_202506.xlsx',
    '/content/drive/MyDrive/Data PT Check/202507/merged_202507.xlsx',
    '/content/drive/MyDrive/Data PT Check/202508/merged_202508.xlsx',
    '/content/drive/MyDrive/Data PT Check/202509/merged_202509.xlsx',
    '/content/drive/MyDrive/Data PT Check/202510/merged_202510.xlsx',
    '/content/drive/MyDrive/Data PT Check/202511/merged_202511.xlsx',
     '/content/drive/MyDrive/Data PT Check/202512/merged_202512.xlsx'
]

# Baca semua file menjadi DataFrame
dfs = [pd.read_excel(path) for path in file_paths]

# Gabungkan semua DataFrame
merged_2025_df = pd.concat(dfs, ignore_index=True)

#delete missing kolom dari toko

merged_2025_df = merged_2025_df.dropna(subset=['Toko'])
#===

# Fungsi untuk mengekstrak panjang (length) dalam cm dari kolom Description
def extract_length(description):
    match = re.search(r'length\s*:\s*(\d+)', description, re.IGNORECASE)
    if match:
        return int(match.group(1))
    return None

# Terapkan fungsi ke kolom Description
merged_2025_df['length'] = merged_2025_df['Description'].astype(str).apply(extract_length)


#as 19 juni 17:50 buat variabel Harga per meter

merged_2025_df['Harga_per_meter'] = merged_2025_df['UnitPrice(incPPN)'] / (merged_2025_df['length'] / 100)

#HAPUS KOLOM YANG TIDAK PERLU
# Daftar kolom yang ingin dihapus
kolom_dihapus = [
    '4M.1', '6M.1', '4M.2', '6M.2', '4M.3', '6M.3', '3M.1', 'Unnamed: 28',
    '10010038', 'CYZ-HR-74004', '10', 'D2081', 'Unnamed: 30', 'Unnamed: 31',
    'Unnamed: 32', 'Unnamed: 33', '轻', '3M.2', 'Unnamed: 27', 'Unnamed: 29',
    '4M.4', 'Unnamed: 35', '5M.1', '5M.2', '6M.4', '4M.5', '6M.5',
    'Customer Code', 'FactoryCode-工厂编码.1', 'Picture.1', 'Type.1',
    'Qty/dus.1', 'QtyOrder.1', 'TOTALDUS.1', 'UnitPrice(incPPN).1',
    'Total.1\tRemarks.1', '5M.3',  '3,95 M', '3M', '4M', '5M', '6M', '10010038', '10', 'Total.1', 'Remarks.1'
]

# Menghapus kolom-kolom tersebut jika ada di DataFrame
merged_2025_df = merged_2025_df.drop(columns=[kol for kol in kolom_dihapus if kol in merged_2025_df.columns])

# Simpan kembali ke file (opsional, jika kamu ingin menyimpannya lagi)
merged_2025_df.to_excel(output_path, index=False)


#===

# Simpan hasil gabungan
output_path = '/content/drive/MyDrive/Data PT Check/merged_2025.xlsx'
merged_2025_df.to_excel(output_path, index=False)

print("File berhasil digabung dan disimpan sebagai 'merged_2025.xlsx'")

```
